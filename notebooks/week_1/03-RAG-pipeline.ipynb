### Embedding Function

```python
# Function : Return embedding for a given text using OpenAI's embedding API. 
def get_embedding(text, model="text-embedding-3-small"):
    response = openai.embeddings.create(
        input=text,
        model=model
    )
    return response.data[0].embedding
```
